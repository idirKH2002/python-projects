## analyse syntaxique :


```python
import stanza
nlpStanza=stanza.Pipeline("fr")

def analyseSyntaxe (text):
    st = nlpStanza(text)
    for sent in st.sentences:
        print ("\n".join(str(x.id)+"\t"+x.text+"\t"+x.lemma+"\t"+x.pos+"\t"+str(x.head)+"("+sent.words[x.head-1].text+")\t"+x.deprel for x in sent.words))
texte = """
Bonjour, Monsieur Madame,
J'ai un velo elitique proform modele prevel 75008.0 serie w13n00138.
Je n'ai plus de résistance, j'ai démonté le moteur de résistance.
Est ce que cela peut être ça la panne, ou il y aurait d'autre possibilité de panne
Je souhaiterais commander la pièce mais j'aimerais être sur.
Pour info au niveau de la console tout fonctionne sauf la dureté de la poulie.
Vous remerciant par avance de la réponse que vous pourrez me faire.
Mon adresse mail est la suivante: thierryolympie@hotmail.fr
Vous en souhaitant bonne réception.
Cordialemnt
Thierry OLYMPIE
"""       
analyseSyntaxe(texte)
```

### Segmentation par token


```python
with open("cocorep-velo.txt", encoding="UTF8") as entree:
    fichier=entree.read()
    st = nlpStanza(fichier)
for sent in st.sentences:
    print (" // ".join(x.text for x in sent.tokens))
```

## Description des donné global :



```python
import stanza
from collections import Counter
nlp = stanza.Pipeline('fr')
with open("cocorep-velo.txt", encoding="UTF8") as entree:
    fichier=entree.read()
doc = nlp(fichier)
total_sentences = len(doc.sentences)
total_words = 0
nouns = []
adjectives = []
verbs = []

# Traiter chaque phrase
for sentence in doc.sentences:
    for word in sentence.words:
        total_words += 1
        if word.upos == "NOUN":
            nouns.append(word.lemma)
        elif word.upos == "ADJ":
            adjectives.append(word.lemma)
        elif word.upos in {"VERB", "AUX"}:
            verbs.append(word.lemma)

# Calculs des statistiques
average_sentences_per_message = total_sentences  # Ici, un seul message
average_words_per_sentence = total_words / total_sentences
nouns_count = len(nouns)
adjectives_count = len(adjectives)
verbs_count = len(verbs)

# Lemmes les plus fréquents
most_common_nouns = Counter(nouns).most_common(5)
most_common_adjectives = Counter(adjectives).most_common(5)
most_common_verbs = Counter(verbs).most_common(5)

# Afficher les résultats
print(f"Nombre de phrases : {total_sentences}")
print(f"Nombre moyen de phrases par message : {average_sentences_per_message:.2f}")
print(f"Nombre de mots : {total_words}")
print(f"Nombre moyen de mots par phrase : {average_words_per_sentence:.2f}")
print(f"Nombre de noms communs (NOUN) : {nouns_count}")
print(f"Lemmes des 5 noms communs les plus fréquents : {most_common_nouns}")
print(f"Nombre d'adjectifs (ADJ) : {adjectives_count}")
print(f"Lemmes des 5 adjectifs les plus fréquents : {most_common_adjectives}")
print(f"Nombre de verbes (VERB | AUX) : {verbs_count}")
print(f"Lemmes des 5 verbes les plus fréquents : {most_common_verbs}")
```
